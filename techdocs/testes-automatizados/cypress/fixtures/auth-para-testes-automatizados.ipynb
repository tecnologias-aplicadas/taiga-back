# Obter Token
Este arquivo explica como obter um token de usuário para utilizar na API.

Importante se lembrar que antes de realizar os passos abaixo, é necessário ter o usuário cadastrado. Este cadastro tem que ser feito via interface ou Django Admin, no qual o usuário deverá ter um e-mail real (fake) associado para ser vinculado aos projetos de testes. 

### Passos para gerar o token:
1. Execute o primeiro código e verifique se o path informado aponta para a pasta raiz do projeto, caso aponte para outra pasta o próximo código não irá funcionar.
2. Execute o próximo código. O token do usuário será exibido como output do código executado.
3. Por fim, será criado um arquivo contendo todos os tokens chamado: `tokens.json`


```python
from pathlib import Path
import os
import sys

# Caminho atual
caminho_atual = os.getcwd()

# Sobe 4 níveis até chegar no diretório raíz do projeto
caminho_techdocs = os.path.abspath(os.path.join(caminho_atual, '..', '..', '..','..'))
# Caminho absoluto do projeto 
sys.path.append(caminho_techdocs)  # Adiciona o caminho do projeto para o Jupyter conseguir "enxergar a pasta ditas como módulos"
PROJECT_ROOT =Path(caminho_techdocs)

# Obtém o diretório atual onde o notebook está sendo executado
#PROJECT_ROOT = Path(os.getcwd())


# Obtém o nome da pasta atual
current_folder = os.path.basename(PROJECT_ROOT)

# Verifica se a pasta atual é 'TAIGA-BACK'
if current_folder == "techdocs":
    # Sobe um diretório
    # Define o diretório de trabalho como UM nível acima
    parent_dir = PROJECT_ROOT.parent
    os.chdir(parent_dir)
    print(f"Subiu para o diretório: {os.getcwd()}")
else:
    print(f"Permanecendo no diretório: {PROJECT_ROOT}")

print(f"Diretório raiz fixado como: {PROJECT_ROOT}")


# Nome do arquivo onde será salvo os tokens do projeto especificado
# Agora use o operador '/' para concatenar caminhos
json_file = PROJECT_ROOT / 'techdocs' / 'testes-automatizados' / 'cypress' / 'fixtures' / 'tokens.json'
```

```python

# Lista de usuários com email e senha
_users = [
    {"username": "usuario-qa", "password": "<senha>"},
    {"username": "usuario-qa1", "password": "<senha>"},
    {"username": "usuario-qa2", "password": "<senha>"}
]
```

## Função que gera os tokens 
Cria um vetor de tokens.

```python
import requests

def login():
    tokens = []

    for user in _users:
        body = {
            "username": user["username"],
            "password": user["password"],
            "g-recaptcha-response": "",
            "type": "normal"
        }
        
        try:
            response = requests.post("http://localhost:8000/api/v1/auth", json=body)
            #response = requests.post("http://<host-do-ambiente-de-testes>/api/v1/auth", json=body)
            response.raise_for_status()  # Lança erro se não for 200
            data = response.json()
            token = data.get("auth_token")
            tokens.append(token)
            print(f"Usuário {user['username']} -> Token: {token}")
        except requests.exceptions.RequestException as e:
            print(f"Erro ao fazer login para {user['username']}: {e}")
            tokens.append(None)

    return tokens

# Exemplo de uso
if __name__ == "__main__":
    tokens = login()
    print("Tokens:", tokens) # Saída: Tokens: [ey, ey, ey, ey]

```

## Salva os tokens

```python

import json

print("tamanho: "+str(len(tokens))) 


# Criar uma lista no formato desejado
data = [{"token": token} for token in tokens] 


# Escrever no arquivo JSON
with open(json_file, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=4)

print(f"Tokens salvos em {json_file}")

```

## Migrations

```python
# Para subir alterações para o banco de dados
DJANGO_SETTINGS_MODULE=settings.config python manage.py migrate --noinput

# Para criar novas migrations em modelos específicos
DJANGO_SETTINGS_MODULE=settings.config python manage.py makemigrations card_relation

# PARA REVERTER TODAS AS MIGRAÇÕES ATÉ A 0001
DJANGO_SETTINGS_MODULE=settings.config python manage.py migrate card_relations zero

DJANGO_SETTINGS_MODULE=settings.config python manage.py makemigrations card_relations
DJANGO_SETTINGS_MODULE=settings.config python manage.py migrate card_relations
DJANGO_SETTINGS_MODULE=settings.config python manage.py migrate 0002
```
